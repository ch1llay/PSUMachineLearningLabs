# Градиентный спуск — визуальная демонстрация

Интерактивные графики на Plotly, показывающие:

1. Функцию ошибки и её производную в разных точках
2. Пошаговый градиентный спуск в 1D
3. Влияние learning rate (слишком маленький / нормальный / слишком большой)
4. Градиент в 2D: векторное поле и траектория спуска

````python
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots
````

## 1. Функция и её производная

Возьмём параболу **L(w) = (w − 3)²**. Её минимум очевидно в `w=3`. Производная: **L'(w) = 2·(w − 3)**.

Построим график функции и касательные в трёх точках. Наклон касательной = значение производной в этой точке. Знак производной сразу виден по направлению касательной.

````python
def L(w):
    return (w - 3) ** 2

def dL(w):
    return 2 * (w - 3)

w_grid = np.linspace(-2, 8, 200)

fig = make_subplots(
    rows=2, cols=1, shared_xaxes=True, vertical_spacing=0.08,
    subplot_titles=("Функция L(w) и касательные", "Производная L′(w) = 2·(w−3)"),
    row_heights=[0.6, 0.4],
)

# --- Верх: функция + касательные ---
fig.add_trace(
    go.Scatter(x=w_grid, y=L(w_grid), mode="lines", name="L(w) = (w−3)²",
               line=dict(color="steelblue", width=3)),
    row=1, col=1,
)

points = [
    (-1, "red",    "Слева от минимума"),
    (3,  "green",  "В минимуме"),
    (6,  "orange", "Справа от минимума"),
]
for w0, color, label in points:
    slope = dL(w0)
    tangent_x = np.linspace(w0 - 2, w0 + 2, 50)
    tangent_y = L(w0) + slope * (tangent_x - w0)
    fig.add_trace(
        go.Scatter(x=tangent_x, y=tangent_y, mode="lines",
                   name=f"{label}: w={w0}, L′={slope:+.0f}",
                   line=dict(color=color, width=2, dash="dash")),
        row=1, col=1,
    )
    fig.add_trace(
        go.Scatter(x=[w0], y=[L(w0)], mode="markers",
                   marker=dict(color=color, size=12, line=dict(color="black", width=1)),
                   showlegend=False),
        row=1, col=1,
    )

# --- Низ: график производной ---
fig.add_trace(
    go.Scatter(x=w_grid, y=dL(w_grid), mode="lines", name="L′(w)",
               line=dict(color="purple", width=3)),
    row=2, col=1,
)

# Заливка: где производная < 0 (функция убывает) и где > 0 (растёт)
fig.add_trace(
    go.Scatter(x=w_grid, y=np.where(dL(w_grid) < 0, dL(w_grid), 0),
               mode="lines", fill="tozeroy", name="L′<0: функция убывает",
               line=dict(color="rgba(0,150,0,0)"),
               fillcolor="rgba(0,200,0,0.15)"),
    row=2, col=1,
)
fig.add_trace(
    go.Scatter(x=w_grid, y=np.where(dL(w_grid) > 0, dL(w_grid), 0),
               mode="lines", fill="tozeroy", name="L′>0: функция растёт",
               line=dict(color="rgba(200,0,0,0)"),
               fillcolor="rgba(200,0,0,0.15)"),
    row=2, col=1,
)

# Горизонтальная ось y=0
fig.add_hline(y=0, line=dict(color="black", width=1), row=2, col=1)

# Те же три точки на графике производной
for w0, color, _ in points:
    fig.add_trace(
        go.Scatter(x=[w0], y=[dL(w0)], mode="markers+text",
                   text=[f"L′={dL(w0):+.0f}"], textposition="top center",
                   marker=dict(color=color, size=12, line=dict(color="black", width=1)),
                   showlegend=False),
        row=2, col=1,
    )

fig.update_xaxes(title_text="w (параметр)", row=2, col=1)
fig.update_yaxes(title_text="L(w)", row=1, col=1)
fig.update_yaxes(title_text="L′(w)", row=2, col=1)
fig.update_layout(height=750, template="plotly_white",
                  title="Функция и её производная")
fig.show()
````

**Что видно:**
- В `w=-1`: касательная идёт вниз-направо. Производная отрицательная (−8). Значит функция убывает → надо идти **вправо** (увеличить w)
- В `w=3`: касательная горизонтальная. Производная = 0. Это минимум, стоп
- В `w=6`: касательная идёт вверх-направо. Производная положительная (+6). Функция растёт → надо идти **влево** (уменьшить w)

Правило `w_new = w − lr · L'(w)` автоматически делает правильный шаг в обоих случаях.

## 2. Градиентный спуск пошагово

Стартуем из `w=−1` (слева от минимума), делаем 15 шагов с `lr=0.2`. Покажем каждый шаг стрелкой.

````python
w = -1.0
lr = 0.2
history = [w]
for _ in range(15):
    grad = dL(w)
    w = w - lr * grad
    history.append(w)

history = np.array(history)

fig = go.Figure()
fig.add_trace(go.Scatter(x=w_grid, y=L(w_grid), mode='lines', name='L(w)',
                         line=dict(color='steelblue', width=3)))

fig.add_trace(go.Scatter(x=history, y=L(history), mode='markers+lines+text',
                         name='Шаги спуска',
                         marker=dict(color='red', size=10, line=dict(color='black', width=1)),
                         line=dict(color='red', width=1, dash='dot'),
                         text=[f'{i}' for i in range(len(history))],
                         textposition='top center', textfont=dict(size=10)))

fig.add_trace(go.Scatter(x=[3], y=[0], mode='markers', name='Минимум',
                         marker=dict(color='green', size=15, symbol='star',
                                     line=dict(color='black', width=1))))

fig.update_layout(title=f'Градиентный спуск: старт w=−1, lr={lr}, 15 шагов',
                  xaxis_title='w', yaxis_title='L(w)',
                  height=500, template='plotly_white')
fig.show()

print('Траектория w:')
for i, w_val in enumerate(history):
    print(f'  шаг {i:2d}: w = {w_val:+.4f}, L(w) = {L(w_val):.4f}, L\'(w) = {dL(w_val):+.4f}')
````

**Замечание:** на первых шагах производная большая по модулю → шаги крупные. По мере приближения к минимуму производная стремится к 0 → шаги становятся всё меньше. Это **автоматическое замедление** — удобное свойство градиентного спуска.

## 3. Влияние learning rate

Сравним три значения lr на одной и той же функции. Стартуем всегда из `w=−1`, делаем 20 шагов.

````python
def gradient_descent(w0, lr, n_steps=20):
    w = w0
    history = [w]
    for _ in range(n_steps):
        w = w - lr * dL(w)
        history.append(w)
    return np.array(history)

configs = [
    (0.05, 'Маленький lr=0.05',  'orange', 'Медленно, не дошли до минимума'),
    (0.2,  'Нормальный lr=0.2',  'green',  'Плавно сходится'),
    (0.95, 'Большой lr=0.95',    'red',    'Колебания около минимума'),
]

fig = make_subplots(rows=1, cols=3, subplot_titles=[c[1] for c in configs])

for col, (lr_val, title, color, _) in enumerate(configs, start=1):
    hist = gradient_descent(-1.0, lr_val)
    fig.add_trace(go.Scatter(x=w_grid, y=L(w_grid), mode='lines',
                             line=dict(color='steelblue', width=2), showlegend=False), row=1, col=col)
    fig.add_trace(go.Scatter(x=hist, y=L(hist), mode='markers+lines',
                             marker=dict(color=color, size=8, line=dict(color='black', width=1)),
                             line=dict(color=color, width=1, dash='dot'),
                             showlegend=False), row=1, col=col)
    fig.add_trace(go.Scatter(x=[3], y=[0], mode='markers',
                             marker=dict(color='green', size=12, symbol='star'),
                             showlegend=False), row=1, col=col)

fig.update_layout(title='Сравнение learning rate', height=450, template='plotly_white')
fig.update_xaxes(title_text='w')
fig.update_yaxes(title_text='L(w)')
fig.show()

for lr_val, title, _, comment in configs:
    hist = gradient_descent(-1.0, lr_val)
    print(f'{title:20s}: w_финал = {hist[-1]:+.4f}, L = {L(hist[-1]):.4f}  ←  {comment}')
````

**Выводы:**
- Слишком маленький `lr` — сходимость медленная, за 20 шагов не дошли
- Нормальный `lr` — быстро и плавно сошлись
- Слишком большой `lr` — перелетаем минимум, колебания вокруг него (если ещё больше, спуск вообще расходится → `L` растёт)

Подбор lr — один из главных гиперпараметров в ML.

## 4. Градиент в 2D

Теперь функция от двух переменных: **L(w₁, w₂) = (w₁ − 3)² + (w₂ + 2)²**. Минимум в точке (3, −2).

Частные производные:
- ∂L/∂w₁ = 2·(w₁ − 3)
- ∂L/∂w₂ = 2·(w₂ + 2)

Градиент `∇L = (∂L/∂w₁, ∂L/∂w₂)` — это вектор. Покажем contour plot + стрелки градиента.

````python
def L2(w1, w2):
    return (w1 - 3) ** 2 + (w2 + 2) ** 2

def grad_L2(w1, w2):
    return np.array([2 * (w1 - 3), 2 * (w2 + 2)])

w1_grid = np.linspace(-3, 8, 100)
w2_grid = np.linspace(-7, 3, 100)
W1, W2 = np.meshgrid(w1_grid, w2_grid)
Z = L2(W1, W2)

fig = go.Figure()
fig.add_trace(go.Contour(x=w1_grid, y=w2_grid, z=Z, colorscale='Blues', showscale=True,
                         contours=dict(start=0, end=100, size=5), colorbar=dict(title='L')))

arrow_pts = [(0, 0), (6, 0), (0, -5), (6, -5), (1, -3), (5, -1)]
for w1, w2 in arrow_pts:
    g = grad_L2(w1, w2)
    scale = 0.15
    fig.add_annotation(x=w1 - scale * g[0], y=w2 - scale * g[1], ax=w1, ay=w2,
                       xref='x', yref='y', axref='x', ayref='y',
                       showarrow=True, arrowhead=3, arrowsize=1.5, arrowwidth=2, arrowcolor='red')

fig.add_trace(go.Scatter(x=[3], y=[-2], mode='markers+text', text=['минимум'],
                         textposition='top right', name='Минимум (3, −2)',
                         marker=dict(color='green', size=15, symbol='star',
                                     line=dict(color='black', width=1))))

fig.update_layout(title='Contour L(w₁, w₂) + стрелки антиградиента (−∇L) — направления спуска',
                  xaxis_title='w₁', yaxis_title='w₂',
                  height=600, width=700, template='plotly_white')
fig.show()
````

**Что видно:** все красные стрелки (антиградиент) смотрят в сторону минимума. Градиентный спуск = двигаемся шаг за шагом по этим стрелкам.

## 5. Траектория спуска в 2D

Запустим градиентный спуск из точки `(−2, 2)` и нарисуем путь на contour plot.

````python
w = np.array([-2.0, 2.0])
lr = 0.15
path = [w.copy()]
for _ in range(25):
    g = grad_L2(w[0], w[1])
    w = w - lr * g
    path.append(w.copy())
path = np.array(path)

fig = go.Figure()
fig.add_trace(go.Contour(x=w1_grid, y=w2_grid, z=Z, colorscale='Blues',
                         contours=dict(start=0, end=100, size=5),
                         colorbar=dict(title='L')))

fig.add_trace(go.Scatter(x=path[:, 0], y=path[:, 1], mode='markers+lines+text',
                         marker=dict(color='red', size=9, line=dict(color='black', width=1)),
                         line=dict(color='red', width=2),
                         text=[str(i) for i in range(len(path))],
                         textposition='top right', textfont=dict(size=9),
                         name='Траектория'))

fig.add_trace(go.Scatter(x=[3], y=[-2], mode='markers', name='Минимум',
                         marker=dict(color='green', size=15, symbol='star',
                                     line=dict(color='black', width=1))))

fig.update_layout(title=f'Путь градиентного спуска в 2D (старт (−2, 2), lr={lr})',
                  xaxis_title='w₁', yaxis_title='w₂',
                  height=600, width=700, template='plotly_white')
fig.show()

print(f'Старт:  w = {path[0]},  L = {L2(*path[0]):.4f}')
print(f'Финиш:  w = {path[-1]},  L = {L2(*path[-1]):.4f}')
print(f'Цель:   w = [3, -2],    L = 0')
````

## 6. Та же траектория в 3D

Посмотрим на «чашу» сбоку — для интуиции, как функция выглядит в пространстве.

````python
fig = go.Figure()
fig.add_trace(go.Surface(x=w1_grid, y=w2_grid, z=Z, colorscale='Blues', opacity=0.8,
                         showscale=False))

path_z = L2(path[:, 0], path[:, 1])
fig.add_trace(go.Scatter3d(x=path[:, 0], y=path[:, 1], z=path_z,
                           mode='markers+lines',
                           marker=dict(color='red', size=5),
                           line=dict(color='red', width=4),
                           name='Траектория'))

fig.add_trace(go.Scatter3d(x=[3], y=[-2], z=[0], mode='markers',
                           marker=dict(color='green', size=8, symbol='diamond'),
                           name='Минимум'))

fig.update_layout(title='3D поверхность L(w₁, w₂) и траектория спуска',
                  scene=dict(xaxis_title='w₁', yaxis_title='w₂', zaxis_title='L'),
                  height=650, template='plotly_white')
fig.show()
````

## 7. А если переменных три?

Функция `L(w₁, w₂, w₃)` «живёт» в 4D (3 входа + 1 выход) — напрямую её не нарисуешь. Но есть трюки:

- **Изоповерхности** — в 3D пространстве параметров рисуем слои `L = const`. Для параболоида `L = (w₁−3)² + (w₂+2)² + (w₃−1)²` изоповерхности — сферы с центром в минимуме
- **Траектория спуска** в 3D пространстве параметров, цвет точек = значение L
- Для **большего числа переменных** (N > 3) визуализация невозможна — только проекции, PCA, срезы

Запустим спуск на 3D задаче.

````python
def L3(w1, w2, w3):
    return (w1 - 3) ** 2 + (w2 + 2) ** 2 + (w3 - 1) ** 2

def grad_L3(w):
    return np.array([2 * (w[0] - 3), 2 * (w[1] + 2), 2 * (w[2] - 1)])

# Градиентный спуск в 3D параметрах
w = np.array([-2.0, 3.0, -3.0])
lr = 0.12
path3 = [w.copy()]
for _ in range(30):
    w = w - lr * grad_L3(w)
    path3.append(w.copy())
path3 = np.array(path3)
L_values = np.array([L3(*p) for p in path3])

# Сетка для изоповерхностей
W1, W2, W3 = np.mgrid[-3:8:40j, -7:4:40j, -4:6:40j]
Lgrid = (W1 - 3) ** 2 + (W2 + 2) ** 2 + (W3 - 1) ** 2

fig = go.Figure()

# Изоповерхности (сферы разного радиуса вокруг минимума)
fig.add_trace(go.Isosurface(
    x=W1.flatten(), y=W2.flatten(), z=W3.flatten(), value=Lgrid.flatten(),
    isomin=5, isomax=50, surface_count=4, opacity=0.15,
    colorscale="Blues", showscale=True, colorbar=dict(title="L", x=1.1),
    name="Изоповерхности L=const",
))

# Траектория спуска, цвет = значение L
fig.add_trace(go.Scatter3d(
    x=path3[:, 0], y=path3[:, 1], z=path3[:, 2],
    mode="markers+lines",
    marker=dict(size=5, color=L_values, colorscale="Reds",
                showscale=True, colorbar=dict(title="L на шаге", x=1.25),
                line=dict(color="black", width=1)),
    line=dict(color="red", width=4),
    name="Траектория",
))

# Минимум
fig.add_trace(go.Scatter3d(
    x=[3], y=[-2], z=[1], mode="markers+text",
    marker=dict(color="green", size=10, symbol="diamond", line=dict(color="black", width=1)),
    text=["минимум (3,−2,1)"], textposition="top center",
    name="Минимум",
))

# Старт
fig.add_trace(go.Scatter3d(
    x=[path3[0, 0]], y=[path3[0, 1]], z=[path3[0, 2]], mode="markers+text",
    marker=dict(color="black", size=8, symbol="circle"),
    text=["старт"], textposition="top center",
    name="Старт",
))

fig.update_layout(
    title="Градиентный спуск в 3D пространстве параметров<br>"
          "<sup>изоповерхности L=const (вложенные сферы) + красная траектория</sup>",
    scene=dict(xaxis_title="w₁", yaxis_title="w₂", zaxis_title="w₃",
               aspectmode="cube"),
    height=700, template="plotly_white",
)
fig.show()

print(f"Старт:  w = {path3[0]},  L = {L_values[0]:.4f}")
print(f"Финиш:  w = {path3[-1].round(3)},  L = {L_values[-1]:.4f}")
print(f"Цель:   w = [3, -2, 1],  L = 0")
````

### Альтернатива: срезы

Ещё способ — зафиксировать одну переменную и рисовать 2D contour. Так видно «разрез» функции в разных плоскостях.

````python
from plotly.subplots import make_subplots

# Фиксируем w3 на разных значениях и рисуем contour по (w1, w2)
w3_fixed_values = [-3, 0, 1, 3]
w1_r = np.linspace(-3, 8, 80)
w2_r = np.linspace(-7, 3, 80)
W1r, W2r = np.meshgrid(w1_r, w2_r)

fig = make_subplots(rows=1, cols=len(w3_fixed_values),
                    subplot_titles=[f"срез при w₃={v}" for v in w3_fixed_values])

for col, w3_val in enumerate(w3_fixed_values, start=1):
    Z = (W1r - 3) ** 2 + (W2r + 2) ** 2 + (w3_val - 1) ** 2
    fig.add_trace(
        go.Contour(x=w1_r, y=w2_r, z=Z, colorscale="Blues",
                   contours=dict(start=0, end=100, size=10),
                   showscale=(col == len(w3_fixed_values))),
        row=1, col=col,
    )
    # Минимум функции в этом срезе (проекция (3, -2))
    fig.add_trace(
        go.Scatter(x=[3], y=[-2], mode="markers",
                   marker=dict(color="green", size=12, symbol="star",
                               line=dict(color="black", width=1)),
                   showlegend=False),
        row=1, col=col,
    )

fig.update_xaxes(title_text="w₁")
fig.update_yaxes(title_text="w₂")
fig.update_layout(
    title="Срезы L(w₁, w₂, w₃) при разных фиксированных w₃ "
          "— минимум на срезе w₃=1 (истинный минимум)",
    height=400, template="plotly_white",
)
fig.show()

print("Видно: минимальное значение L на каждом срезе зависит от w₃.")
print("Только при w₃=1 срез касается истинного минимума L=0.")
print("Это «разрезание» 4D поверхности плоскостями в 3D пространстве.")
````

### Почему больше 3 переменных не нарисовать

- 1 переменная → график y=f(x) в 2D
- 2 переменные → поверхность в 3D (или contour в 2D)
- 3 переменные → изоповерхности в 3D (уже трюк — показываем только слои)
- 4+ переменных → **никак напрямую**. Варианты: PCA/t-SNE проекции, попарные срезы, графики loss vs эпоха

В настоящих ML-моделях параметров **миллионы**. Мы не видим ландшафт функции ошибки — доверяем градиентному спуску «на ощупь», смотрим только на loss и метрики во времени.

## 8. Ландшафт функции потерь для разных моделей

Покажем на 4 примерах, как меняется форма loss-функции в зависимости от того, что стоит между весами и выходом. Все модели — **с двумя параметрами** `(w, b)`, чтобы можно было нарисовать поверхность в 3D и contour сверху.

**Что будем сравнивать:**

| # | Модель | Loss | Ожидаемая форма |
|---|---|---|---|
| 1 | `y = w·x + b` | MSE | Чаша (параболоид) |
| 2 | `y = σ(w·x + b)` | MSE | С плато, нелинейная |
| 3 | `y = σ(w₁·x) + σ(w₂·x)` (mini-MLP) | MSE | Два минимума (симметрия) |
| 4 | `y = σ(w·x + b)` | log-loss | Выпуклая, но не квадратичная |

````python
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots

np.random.seed(42)

# --- Данные для регрессии (случай 1): y = 2x + 1 + шум ---
x_reg = np.linspace(-2, 2, 40)
y_reg = 2 * x_reg + 1 + 0.3 * np.random.randn(40)

# --- Данные для классификации (случаи 2, 4): 1D-бинарная задача ---
x_cls = np.linspace(-3, 3, 60)
y_cls = (x_cls > 0).astype(float)
flip_idx = np.random.choice(60, 4, replace=False)
y_cls[flip_idx] = 1 - y_cls[flip_idx]  # немного шумим метки

def sigmoid(z):
    return 1.0 / (1.0 + np.exp(-np.clip(z, -500, 500)))

def plot_landscape(w_vals, b_vals, L, title, w_label="w", b_label="b",
                   minima=None, log_z=False):
    Z = np.log1p(L) if log_z else L
    fig = make_subplots(
        rows=1, cols=2,
        specs=[[{"type": "xy"}, {"type": "scene"}]],
        subplot_titles=("Вид сверху (contour)", "3D поверхность"),
        horizontal_spacing=0.08,
    )
    fig.add_trace(
        go.Contour(x=w_vals, y=b_vals, z=Z, colorscale="Viridis",
                   ncontours=25, colorbar=dict(x=0.46, title="L")),
        row=1, col=1,
    )
    if minima:
        fig.add_trace(
            go.Scatter(x=[m[0] for m in minima], y=[m[1] for m in minima],
                       mode="markers",
                       marker=dict(color="red", size=14, symbol="star",
                                   line=dict(color="white", width=2)),
                       name="Минимум", showlegend=False),
            row=1, col=1,
        )
    fig.add_trace(
        go.Surface(x=w_vals, y=b_vals, z=Z, colorscale="Viridis",
                   showscale=False, opacity=0.9),
        row=1, col=2,
    )
    fig.update_xaxes(title_text=w_label, row=1, col=1)
    fig.update_yaxes(title_text=b_label, row=1, col=1)
    fig.update_layout(
        title=title, height=520, template="plotly_white",
        scene=dict(xaxis_title=w_label, yaxis_title=b_label, zaxis_title="L"),
    )
    return fig

print(f"Регрессионные данные: {len(x_reg)} точек, y = 2x + 1 + шум")
print(f"Классификационные данные: {len(x_cls)} точек, бинарные метки")
````

### Случай 1: Линейная регрессия + MSE → **чаша**

Формула модели: `y_pred = w·x + b`
Формула loss: `L(w, b) = mean((w·x + b − y)²)`

Функция строго квадратична по `w` и `b` → параболоид → один глобальный минимум.

````python
w_grid = np.linspace(-2, 6, 80)
b_grid = np.linspace(-3, 5, 80)
W, B = np.meshgrid(w_grid, b_grid)

# Векторизованный расчёт: L[i,j] = mean((w_j*x + b_i - y)^2)
y_pred = W[..., None] * x_reg + B[..., None]        # (80, 80, 40)
L1 = ((y_pred - y_reg) ** 2).mean(axis=-1)

# Истинный минимум (для отметки)
w_opt = ((x_reg - x_reg.mean()) * (y_reg - y_reg.mean())).sum() / ((x_reg - x_reg.mean()) ** 2).sum()
b_opt = y_reg.mean() - w_opt * x_reg.mean()
print(f"Минимум: w={w_opt:.3f}, b={b_opt:.3f} (истинные: w=2, b=1)")

fig = plot_landscape(w_grid, b_grid, L1,
                     "Случай 1: Линейная регрессия + MSE — параболоид (одна чаша)",
                     minima=[(w_opt, b_opt)])
fig.show()
````

### Случай 2: Сигмоида + MSE → **нелинейный ландшафт с плато**

Формула модели: `y_pred = σ(w·x + b)`
Формула loss: `L(w, b) = mean((σ(w·x + b) − y)²)`

Теперь между весами и выходом стоит **сигмоида**. Квадратичность пропадает. Поверхность:
- имеет **плато** (там, где сигмоида насыщается — производные близки к 0)
- ограничена сверху (σ ∈ [0, 1], поэтому MSE ≤ 1)
- всё ещё имеет глобальный минимум, но градиенты в плоских областях маленькие → обучение замедляется

````python
w_grid2 = np.linspace(-5, 10, 100)
b_grid2 = np.linspace(-8, 8, 100)
W2, B2 = np.meshgrid(w_grid2, b_grid2)

y_pred2 = sigmoid(W2[..., None] * x_cls + B2[..., None])
L2 = ((y_pred2 - y_cls) ** 2).mean(axis=-1)

i, j = np.unravel_index(L2.argmin(), L2.shape)
w_min2, b_min2 = w_grid2[j], b_grid2[i]
print(f"Минимум на сетке: w={w_min2:.2f}, b={b_min2:.2f}, L={L2.min():.4f}")

fig = plot_landscape(w_grid2, b_grid2, L2,
                     "Случай 2: Сигмоида + MSE — плато + нелинейность",
                     minima=[(w_min2, b_min2)])
fig.show()
````

### Отступление: почему σ(w₁·x) + σ(w₂·x) ≠ σ((w₁+w₂)·x)

Сигмоида **не аддитивна**: `σ(a) + σ(b) ≠ σ(a+b)`.

Проверим на числах: `σ(1) + σ(3) = 0.731 + 0.953 = 1.684`, но `σ(1+3) = σ(4) = 0.982`. Разные числа.

Это **не баг, а фича**. Именно благодаря неаддитивности сумма двух сигмоид создаёт формы, недоступные одной сигмоиде. Если бы σ была аддитивна, два нейрона свернулись бы в один, и нейросети не имели бы смысла.

````python
# Сравнение: одна сигмоида vs сумма двух сигмоид
x_demo = np.linspace(-4, 4, 200)

fig = make_subplots(rows=1, cols=2,
                    subplot_titles=(
                        "Одна сигмоида σ(w·x) — всегда монотонная",
                        "Сумма σ(w₁·x) + σ(w₂·x) — может быть не монотонной",
                    ))

# Слева: одна сигмоида при разных w
for w in [1, 2, 5]:
    fig.add_trace(go.Scatter(x=x_demo, y=sigmoid(w * x_demo),
                             mode="lines", name=f"σ({w}·x)"),
                  row=1, col=1)

# Справа: сумма двух сигмоид — разные интересные формы
combos = [
    (2, -3,  "σ(2·x) + σ(-3·x) — горб сверху"),
    (-2, 3,  "σ(-2·x) + σ(3·x) — тот же горб (симметрия!)"),
    (5, 5,   "σ(5·x) + σ(5·x) — просто удвоенная"),
    (3, -1,  "σ(3·x) + σ(-1·x) — S-образная с плечом"),
]
for w1, w2, label in combos:
    y = sigmoid(w1 * x_demo) + sigmoid(w2 * x_demo)
    fig.add_trace(go.Scatter(x=x_demo, y=y, mode="lines", name=label),
                  row=1, col=2)

# Пунктирная линия: что было бы, если бы σ была аддитивной (σ((w1+w2)·x))
y_if_additive = sigmoid((2 + (-3)) * x_demo)
fig.add_trace(go.Scatter(x=x_demo, y=y_if_additive, mode="lines",
                         name="σ((2-3)·x) = σ(-x) — если бы σ была аддитивной",
                         line=dict(dash="dash", color="gray", width=2)),
              row=1, col=2)

fig.update_xaxes(title_text="x")
fig.update_yaxes(title_text="y")
fig.update_layout(height=500, template="plotly_white",
                  title="Одна сигмоида не может то, что может сумма двух")
fig.show()

# Численная проверка неаддитивности
print("Проверка неаддитивности сигмоиды:")
for a, b in [(1, 3), (-2, 2), (0.5, 0.5)]:
    lhs = sigmoid(a) + sigmoid(b)
    rhs = sigmoid(a + b)
    print(f"  σ({a}) + σ({b}) = {lhs:.4f}   vs   σ({a}+{b}) = {rhs:.4f}   разница = {lhs-rhs:+.4f}")
````

### Случай 3: Мини-перцептрон (2 нейрона) + MSE → **множественные минимумы**

Формула модели: `y_pred = σ(w₁·x) + σ(w₂·x)` — два скрытых нейрона
Формула loss: `L(w₁, w₂) = mean((y_pred − y)²)`

Цель: `y = σ(2·x) + σ(−3·x)`. Минимум достигается в двух точках: `(w₁=2, w₂=−3)` **и** `(w₁=−3, w₂=2)` — из-за **перестановочной симметрии**: если поменять нейроны местами, выход не изменится.

Это уже **не выпуклая** функция: между двумя минимумами — «горный перевал». Градиентный спуск придёт **в один из двух** минимумов в зависимости от инициализации.

````python
x_mlp = np.linspace(-2, 2, 40)
y_mlp = sigmoid(2 * x_mlp) + sigmoid(-3 * x_mlp)  # целевая функция

w1_grid = np.linspace(-6, 6, 100)
w2_grid = np.linspace(-6, 6, 100)
W1g, W2g = np.meshgrid(w1_grid, w2_grid)

y_pred3 = sigmoid(W1g[..., None] * x_mlp) + sigmoid(W2g[..., None] * x_mlp)
L3 = ((y_pred3 - y_mlp) ** 2).mean(axis=-1)

fig = plot_landscape(w1_grid, w2_grid, L3,
                     "Случай 3: Мини-MLP + MSE — два эквивалентных минимума",
                     w_label="w₁", b_label="w₂",
                     minima=[(2, -3), (-3, 2)])
fig.show()

print("Два красных минимума — одна и та же модель после перестановки нейронов.")
print("Между ними есть седловая точка — градиент = 0, но это не минимум.")
````

### Случай 4: Сигмоида + log-loss → **выпуклая, но не квадратичная**

Формула модели: `y_pred = σ(w·x + b)` (та же, что в случае 2)
Формула loss: `L(w, b) = −mean(y·log(p) + (1−y)·log(1−p))`

Та же сигмоида, но **другая функция потерь**. Теория говорит, что эта комбинация **выпуклая** (одна чаша), хотя и не квадратичная.

Обрати внимание: при очень плохих предсказаниях `log(p → 0) → −∞` → loss уходит в бесконечность. Но **выпуклость сохраняется**, градиенты не затухают на плато (в отличие от MSE + sigmoid).

Это главная причина, почему в классификации используют log-loss, а не MSE.

````python
y_pred4 = sigmoid(W2[..., None] * x_cls + B2[..., None])
eps = 1e-7
L4 = -(y_cls * np.log(y_pred4 + eps) + (1 - y_cls) * np.log(1 - y_pred4 + eps)).mean(axis=-1)

i, j = np.unravel_index(L4.argmin(), L4.shape)
w_min4, b_min4 = w_grid2[j], b_grid2[i]
print(f"Минимум на сетке: w={w_min4:.2f}, b={b_min4:.2f}, L={L4.min():.4f}")

fig = plot_landscape(w_grid2, b_grid2, L4,
                     "Случай 4: Сигмоида + log-loss — выпуклая, но растёт в бесконечность",
                     minima=[(w_min4, b_min4)], log_z=True)
fig.show()
````

### Сравнение всех четырёх ландшафтов бок о бок

Один график — четыре contour'а. Так сразу видно разницу формы.

````python
fig = make_subplots(
    rows=1, cols=4,
    subplot_titles=(
        "1. Linear + MSE<br><sup>параболоид</sup>",
        "2. Sigmoid + MSE<br><sup>плато</sup>",
        "3. MLP + MSE<br><sup>2 минимума</sup>",
        "4. Sigmoid + log-loss<br><sup>выпуклая</sup>",
    ),
    horizontal_spacing=0.07,
)

cases = [
    (w_grid,  b_grid,  L1,                "w",  "b",  [(w_opt, b_opt)]),
    (w_grid2, b_grid2, L2,                "w",  "b",  [(w_min2, b_min2)]),
    (w1_grid, w2_grid, L3,                "w1", "w2", [(2, -3), (-3, 2)]),
    (w_grid2, b_grid2, np.log1p(L4),      "w",  "b",  [(w_min4, b_min4)]),
]

for col, (wv, bv, Z, wl, bl, mins) in enumerate(cases, start=1):
    fig.add_trace(
        go.Contour(x=wv, y=bv, z=Z, colorscale="Viridis", ncontours=20, showscale=False),
        row=1, col=col,
    )
    fig.add_trace(
        go.Scatter(x=[m[0] for m in mins], y=[m[1] for m in mins], mode="markers",
                   marker=dict(color="red", size=14, symbol="star",
                               line=dict(color="white", width=2)),
                   showlegend=False),
        row=1, col=col,
    )
    fig.update_xaxes(title_text=wl, row=1, col=col)
    fig.update_yaxes(title_text=bl, row=1, col=col)

fig.update_layout(
    title="Сравнение 4 ландшафтов loss-функции (красная звезда = минимум)",
    height=450, template="plotly_white",
)
fig.show()
````

### Что увидели

| Случай | Форма | Минимумов | Выпуклость | Подходит для обучения? |
|---|---|---|---|---|
| 1. Linear + MSE | Параболоид | 1 | Выпуклая | Отлично |
| 2. Sigmoid + MSE | Плато + холм | 1 | Невыпуклая (плато душит градиенты) | Медленно |
| 3. MLP + MSE | Много минимумов | ≥ 2 | Сильно невыпуклая | Работает, но не гарантированно |
| 4. Sigmoid + log-loss | Крутая чаша | 1 | Выпуклая | Отлично для классификации |

**Главное наблюдение:** форму ландшафта определяет **комбинация модель × loss**, а не loss сама по себе. Одна и та же сигмоида даёт совершенно разный ландшафт с MSE (плоский, плохой для обучения) и с log-loss (крутая чаша, отличный для обучения).

## 9. Перцептрон как универсальный аппроксиматор

**Теорема универсальной аппроксимации** (Цыбенко, 1989): MLP с **одним скрытым слоем** и достаточным числом нейронов с сигмоидной активацией может приблизить любую непрерывную функцию на ограниченном отрезке с любой точностью.

Сравним с рядом Тейлора:

| | Ряд Тейлора | MLP |
|---|---|---|
| Базис | степени `xⁿ` | сигмоиды `σ(wx+b)` |
| Область | локальная | глобальная |
| Улучшаем точность | увеличиваем степень | увеличиваем число нейронов |

Обучим MLP приближать `sin(x)` на отрезке `[−2π, 2π]` с разным числом нейронов. Чем больше нейронов — тем точнее.

````python
from sklearn.neural_network import MLPRegressor

# Целевая функция: синус на [-2π, 2π]
x_train = np.linspace(-2 * np.pi, 2 * np.pi, 300).reshape(-1, 1)
y_train = np.sin(x_train).ravel()

# Обучим 4 модели с разным числом нейронов
n_neurons_list = [1, 3, 10, 50]
models = {}

for n in n_neurons_list:
    model = MLPRegressor(
        hidden_layer_sizes=(n,),
        activation="logistic",      # сигмоида
        solver="lbfgs",             # быстрый оптимизатор для маленьких сетей
        max_iter=5000,
        random_state=42,
        tol=1e-8,
    )
    model.fit(x_train, y_train)
    y_pred = model.predict(x_train)
    mse = np.mean((y_pred - y_train) ** 2)
    models[n] = (model, y_pred, mse)
    print(f"{n:3d} нейронов: MSE = {mse:.6f}")

# Визуализация
fig = make_subplots(
    rows=2, cols=2,
    subplot_titles=[f"{n} нейрон(ов), MSE={models[n][2]:.4f}" for n in n_neurons_list],
)
for idx, n in enumerate(n_neurons_list):
    row, col = idx // 2 + 1, idx % 2 + 1
    _, y_pred, _ = models[n]
    fig.add_trace(go.Scatter(x=x_train.ravel(), y=y_train, mode="lines",
                             name="sin(x)", line=dict(color="steelblue", width=3),
                             showlegend=(idx == 0)),
                  row=row, col=col)
    fig.add_trace(go.Scatter(x=x_train.ravel(), y=y_pred, mode="lines",
                             name="MLP прогноз", line=dict(color="red", width=2, dash="dash"),
                             showlegend=(idx == 0)),
                  row=row, col=col)

fig.update_xaxes(title_text="x")
fig.update_yaxes(title_text="y")
fig.update_layout(height=700, template="plotly_white",
                  title="Приближение sin(x) с помощью MLP (сигмоида, 1 скрытый слой)")
fig.show()
````

### Разложение на отдельные сигмоиды

MLP с 10 нейронами — это сумма **10 сигмоид** с разными весами/смещениями + константа. Нарисуем каждую отдельно, чтобы увидеть, как они складываются в синус.

````python
# Извлечём параметры MLP с 10 нейронами
n_decomp = 10
model = models[n_decomp][0]

W1 = model.coefs_[0].ravel()    # веса входного слоя → скрытый слой (10 штук)
b1 = model.intercepts_[0]        # смещения скрытых нейронов (10 штук)
W2 = model.coefs_[1].ravel()    # веса скрытого слоя → выход (10 штук)
b2 = model.intercepts_[1][0]     # смещение выходного нейрона

# Каждая компонента: v_i * σ(w_i * x + b_i)
x_fine = np.linspace(-2 * np.pi, 2 * np.pi, 300)

fig = make_subplots(rows=1, cols=2,
                    subplot_titles=("10 отдельных нейронов v_i · σ(w_i·x + b_i)",
                                    "Их сумма + смещение = приближение sin(x)"))

# Слева: каждый нейрон отдельно
for i in range(n_decomp):
    h_i = W2[i] * sigmoid(W1[i] * x_fine + b1[i])
    fig.add_trace(go.Scatter(x=x_fine, y=h_i, mode="lines",
                             name=f"нейрон {i+1}: v={W2[i]:+.2f}",
                             line=dict(width=1.5), showlegend=False),
                  row=1, col=1)

# Справа: сумма + истинный sin
total = b2 + sum(W2[i] * sigmoid(W1[i] * x_fine + b1[i]) for i in range(n_decomp))
fig.add_trace(go.Scatter(x=x_fine, y=np.sin(x_fine), mode="lines",
                         name="sin(x)", line=dict(color="steelblue", width=3)),
              row=1, col=2)
fig.add_trace(go.Scatter(x=x_fine, y=total, mode="lines",
                         name="Сумма всех нейронов", line=dict(color="red", width=2, dash="dash")),
              row=1, col=2)

fig.update_xaxes(title_text="x")
fig.update_yaxes(title_text="y")
fig.update_layout(height=500, template="plotly_white",
                  title=f"Как {n_decomp} сигмоид складываются в приближение sin(x)")
fig.show()

print("Каждая тонкая линия слева — одна сигмоида, сдвинутая и масштабированная.")
print("Их сумма (справа красная пунктирная) приближает синус.")
print("Больше нейронов → ближе к истинному синусу.")
````

### Сравнение с рядом Тейлора

Ряд Тейлора `sin(x)` около 0:

```
sin(x) ≈ x − x³/6 + x⁵/120 − x⁷/5040 + ...
```

Хорошо работает около 0, но на `|x| > π` разваливается — даже с 10 членами ряда. MLP равномерно приближает на всём отрезке.

````python
from math import factorial

x_compare = np.linspace(-2 * np.pi, 2 * np.pi, 300)
y_true = np.sin(x_compare)

# Ряд Тейлора: sin(x) = x - x^3/3! + x^5/5! - x^7/7! + ...
def taylor_sin(x, n_terms):
    result = np.zeros_like(x)
    for k in range(n_terms):
        result += (-1) ** k * x ** (2 * k + 1) / factorial(2 * k + 1)
    return result

# MLP с 10 нейронами (уже обучен выше)
y_mlp10 = models[10][1]

fig = go.Figure()
fig.add_trace(go.Scatter(x=x_compare, y=y_true, mode="lines",
                         name="sin(x) — истина",
                         line=dict(color="steelblue", width=4)))

# Ряды Тейлора разного порядка
for n_terms, color in [(3, "orange"), (5, "green"), (10, "purple")]:
    y_t = taylor_sin(x_compare, n_terms)
    fig.add_trace(go.Scatter(x=x_compare, y=y_t, mode="lines",
                             name=f"Тейлор ({n_terms} членов)",
                             line=dict(color=color, width=1.5, dash="dot")))

# MLP
fig.add_trace(go.Scatter(x=x_train.ravel(), y=y_mlp10, mode="lines",
                         name="MLP (10 нейронов)",
                         line=dict(color="red", width=2, dash="dash")))

fig.update_layout(
    title="Ряд Тейлора vs MLP: приближение sin(x) на [−2π, 2π]",
    xaxis_title="x", yaxis_title="y",
    height=500, template="plotly_white",
    yaxis=dict(range=[-2, 2]),
)
fig.show()

print("Видно: ряд Тейлора хорош около 0, но взрывается на краях.")
print("MLP даёт равномерное приближение по всему отрезку.")
````

### Выводы

- **Да, перцептрон — универсальный аппроксиматор.** MLP с одним скрытым слоем и достаточным числом нейронов с сигмоидной активацией может приблизить любую непрерывную функцию.
- **Синус приближается легко** — 10 нейронов дают приемлемое качество, 50 — отличное.
- **Отличие от ряда Тейлора:** Тейлор использует полиномиальный базис (степени `x`) и работает **локально**; MLP использует сигмоидный базис и работает **глобально**.
- **Платёж за универсальность:** функция потерь при обучении MLP **не выпуклая** (как мы видели в Случае 3). Нужен градиентный спуск с удачной инициализацией и много итераций.

## 10. Фурье vs MLP — одна и та же идея, разные базисы

Сравним две реализации идеи «сложную функцию = сумма простых кубиков»:

- **Ряд Фурье**: кубики = **синусы** `sin(kx), cos(kx)` с фиксированными частотами `k`
- **MLP**: кубики = **сигмоиды** `σ(wx+b)` с обучаемыми `w, b`

Для наглядности возьмём «прямоугольную волну» (square wave) — классический пример из теории Фурье. Обе техники умеют её приближать, но **кубики разные**.

````python
from sklearn.neural_network import MLPRegressor

# Целевая функция: прямоугольная волна
x_target = np.linspace(-2 * np.pi, 2 * np.pi, 400)
y_target = np.sign(np.sin(x_target))  # +1 / -1

# --- Фурье-разложение прямоугольной волны ---
# Классическая формула: f(x) = (4/π) · Σ sin((2k-1)x) / (2k-1) для k=1,2,...
def fourier_square(x, n_terms):
    result = np.zeros_like(x)
    components = []
    for k in range(1, n_terms + 1):
        comp = (4 / np.pi) * np.sin((2 * k - 1) * x) / (2 * k - 1)
        components.append(comp)
        result += comp
    return result, components

N_FOURIER = 5
fourier_sum, fourier_comps = fourier_square(x_target, N_FOURIER)

# --- MLP-разложение ---
N_MLP = 15
mlp = MLPRegressor(hidden_layer_sizes=(N_MLP,), activation="logistic",
                   solver="lbfgs", max_iter=8000, random_state=42, tol=1e-8)
mlp.fit(x_target.reshape(-1, 1), y_target)
mlp_sum = mlp.predict(x_target.reshape(-1, 1))

W1 = mlp.coefs_[0].ravel()
b1 = mlp.intercepts_[0]
W2 = mlp.coefs_[1].ravel()
b2 = mlp.intercepts_[1][0]
mlp_comps = [W2[i] * sigmoid(W1[i] * x_target + b1[i]) for i in range(N_MLP)]

# --- Визуализация ---
fig = make_subplots(
    rows=2, cols=2,
    subplot_titles=(
        f"Фурье — {N_FOURIER} базисных синусов (отдельно)",
        f"Фурье — их сумма приближает прямоугольник",
        f"MLP — {N_MLP} базисных сигмоид (отдельно)",
        f"MLP — их сумма приближает тот же прямоугольник",
    ),
    vertical_spacing=0.12, horizontal_spacing=0.08,
)

# (1,1) Фурье — компоненты
for k, comp in enumerate(fourier_comps):
    fig.add_trace(go.Scatter(x=x_target, y=comp, mode="lines",
                             name=f"sin({2*k+1}x)/{2*k+1}",
                             line=dict(width=1.5), showlegend=False),
                  row=1, col=1)

# (1,2) Фурье — сумма
fig.add_trace(go.Scatter(x=x_target, y=y_target, mode="lines",
                         name="прямоугольник", line=dict(color="steelblue", width=3),
                         showlegend=False),
              row=1, col=2)
fig.add_trace(go.Scatter(x=x_target, y=fourier_sum, mode="lines",
                         name="Фурье", line=dict(color="red", width=2, dash="dash"),
                         showlegend=False),
              row=1, col=2)

# (2,1) MLP — компоненты
for i, comp in enumerate(mlp_comps):
    fig.add_trace(go.Scatter(x=x_target, y=comp, mode="lines",
                             line=dict(width=1.2), showlegend=False),
                  row=2, col=1)

# (2,2) MLP — сумма
fig.add_trace(go.Scatter(x=x_target, y=y_target, mode="lines",
                         line=dict(color="steelblue", width=3), showlegend=False),
              row=2, col=2)
fig.add_trace(go.Scatter(x=x_target, y=mlp_sum, mode="lines",
                         line=dict(color="red", width=2, dash="dash"), showlegend=False),
              row=2, col=2)

fig.update_xaxes(title_text="x")
fig.update_yaxes(title_text="y")
fig.update_layout(height=800, template="plotly_white",
                  title="Два разных базиса приближают одну и ту же прямоугольную волну")
fig.show()

mse_fourier = np.mean((fourier_sum - y_target) ** 2)
mse_mlp = np.mean((mlp_sum - y_target) ** 2)
print(f"MSE Фурье ({N_FOURIER} синусов):  {mse_fourier:.4f}")
print(f"MSE MLP ({N_MLP} сигмоид):       {mse_mlp:.4f}")
````

### Что увидели

**Верхний ряд (Фурье):**
- Слева — **5 синусоид** разной частоты и амплитуды. Каждая — **волна**, сама по себе выглядит красиво.
- Справа — их сумма. Видно «рябь» на плоских участках (эффект Гиббса) — характерный признак Фурье-приближения разрывных функций.

**Нижний ряд (MLP):**
- Слева — **15 сигмоид**. Каждая — **монотонная ступенька**, никаких волн.
- Справа — их сумма приближает тот же прямоугольник, но гораздо ближе к ступенькам (без ряби).

**Ключевое отличие базисов:**

| | Фурье | MLP |
|---|---|---|
| Каждый кубик | Волна `sin(kx)` — красивая, но «не локальная» | Ступенька — локализованный переход |
| Частоты/позиции | **Фиксированы** заранее (k=1,2,3,...) | **Обучаются** под данные |
| Для разрывов | Эффект Гиббса (рябь у скачков) | Ступеньки отлично ложатся на разрывы |
| Для гладких функций | Идеально (пара синусов хватит) | Нужно больше нейронов |

**Вывод:** обе техники — разные способы «сложить сложное из простого». Фурье выбирает **фиксированный базис** (заранее известный и ортогональный). MLP **сам подбирает базис** под задачу — каждая сигмоида адаптируется. Для разрывных/ступенчатых данных это большое преимущество.

### То же самое для треугольной волны

Треугольная волна — непрерывная (без разрывов), но с «изломами» в вершинах. Посмотрим, как её приближают Фурье и MLP.

Формула Фурье для треугольника с периодом 2π и амплитудой 1:

```
f(x) = (8/π²) · Σ (-1)^k / (2k+1)² · sin((2k+1)·x),  k = 0, 1, 2, ...
```

В отличие от прямоугольника, здесь коэффициенты убывают как `1/k²` (а не `1/k`), поэтому Фурье-ряд сходится быстрее.

````python
# Целевая функция: треугольная волна через arcsin(sin(x))
x_t = np.linspace(-2 * np.pi, 2 * np.pi, 400)
y_t = (2 / np.pi) * np.arcsin(np.sin(x_t))

# --- Фурье-разложение треугольной волны ---
def fourier_triangle(x, n_terms):
    result = np.zeros_like(x)
    components = []
    for k in range(n_terms):
        coef = (8 / np.pi ** 2) * ((-1) ** k) / (2 * k + 1) ** 2
        comp = coef * np.sin((2 * k + 1) * x)
        components.append(comp)
        result += comp
    return result, components

N_FOURIER_T = 5
fourier_sum_t, fourier_comps_t = fourier_triangle(x_t, N_FOURIER_T)

# --- MLP на треугольной волне ---
N_MLP_T = 30
mlp_t = MLPRegressor(hidden_layer_sizes=(N_MLP_T,), activation="logistic",
                     solver="lbfgs", max_iter=8000, random_state=42, tol=1e-8)
mlp_t.fit(x_t.reshape(-1, 1), y_t)
mlp_sum_t = mlp_t.predict(x_t.reshape(-1, 1))

W1_t = mlp_t.coefs_[0].ravel()
b1_t = mlp_t.intercepts_[0]
W2_t = mlp_t.coefs_[1].ravel()
b2_t = mlp_t.intercepts_[1][0]
mlp_comps_t = [W2_t[i] * sigmoid(W1_t[i] * x_t + b1_t[i]) for i in range(N_MLP_T)]

# --- Визуализация ---
fig = make_subplots(
    rows=2, cols=2,
    subplot_titles=(
        f"Фурье — {N_FOURIER_T} базисных синусов (отдельно)",
        f"Фурье — сумма приближает треугольник (почти без ряби)",
        f"MLP — {N_MLP_T} базисных сигмоид (отдельно)",
        f"MLP — сумма приближает тот же треугольник",
    ),
    vertical_spacing=0.12, horizontal_spacing=0.08,
)

for comp in fourier_comps_t:
    fig.add_trace(go.Scatter(x=x_t, y=comp, mode="lines",
                             line=dict(width=1.5), showlegend=False),
                  row=1, col=1)

fig.add_trace(go.Scatter(x=x_t, y=y_t, mode="lines",
                         line=dict(color="steelblue", width=3), showlegend=False),
              row=1, col=2)
fig.add_trace(go.Scatter(x=x_t, y=fourier_sum_t, mode="lines",
                         line=dict(color="red", width=2, dash="dash"), showlegend=False),
              row=1, col=2)

for comp in mlp_comps_t:
    fig.add_trace(go.Scatter(x=x_t, y=comp, mode="lines",
                             line=dict(width=1.2), showlegend=False),
                  row=2, col=1)

fig.add_trace(go.Scatter(x=x_t, y=y_t, mode="lines",
                         line=dict(color="steelblue", width=3), showlegend=False),
              row=2, col=2)
fig.add_trace(go.Scatter(x=x_t, y=mlp_sum_t, mode="lines",
                         line=dict(color="red", width=2, dash="dash"), showlegend=False),
              row=2, col=2)

fig.update_xaxes(title_text="x")
fig.update_yaxes(title_text="y")
fig.update_layout(height=800, template="plotly_white",
                  title="Треугольная волна: Фурье vs MLP")
fig.show()

mse_f = np.mean((fourier_sum_t - y_t) ** 2)
mse_m = np.mean((mlp_sum_t - y_t) ** 2)
print(f"MSE Фурье ({N_FOURIER_T} синусов):  {mse_f:.6f}")
print(f"MSE MLP ({N_MLP_T} сигмоид):       {mse_m:.6f}")
````

### Сравнение: прямоугольник vs треугольник

| | Прямоугольник (разрывная) | Треугольник (непрерывная, с изломами) |
|---|---|---|
| Фурье — убывание коэффициентов | `1/k` (медленно) | `1/k²` (быстрее) |
| Рябь Гиббса у Фурье | Заметная | Почти отсутствует |
| MLP — сколько нужно нейронов | Больше (много скачков) | Меньше (плавнее) |

**Вывод:** чем «гладче» функция, тем меньше компонент нужно обоим методам. Фурье очень страдает от **разрывов**, но хорошо справляется с **изломами**. MLP с сигмоидой относительно равномерно приближает и то, и другое.

Для действительно гладких функций (синусы, полиномы) Фурье / Тейлор побеждают по экономичности. Для функций с резкими переходами или сложной структурой (реальные данные в ML!) MLP выигрывает — именно поэтому нейросети стали универсальным инструментом.

### Окружность — случай когда Фурье безоговорочно выигрывает

Окружность — это не функция вида `y = f(x)` (из-за двух ветвей при каждом `x`). Поэтому приближать её надо **параметрически**: на вход подаём `t ∈ [0, 2π]`, на выходе пара координат `(x(t), y(t))`.

Истинная окружность радиуса 1:
```
x(t) = cos(t)
y(t) = sin(t)
```

Интересно: это **в точности** базис Фурье. То есть **Фурье с одним членом даёт идеальную окружность**. А MLP приходится аппроксимировать синус и косинус суммой сигмоид — много работы ради того, что Фурье делает за один вызов.

Это наглядный пример: когда базис **совпадает с задачей**, Фурье непобедим. Когда задача произвольная (реальные данные), выигрывает MLP.

````python
# Параметрическое описание окружности
t = np.linspace(0, 2 * np.pi, 300)
cos_true = np.cos(t)
sin_true = np.sin(t)

# --- Фурье: тривиально, 1 член ---
fourier_x = np.cos(t)    # это и есть "первая гармоника по косинусу"
fourier_y = np.sin(t)    # "первая гармоника по синусу"

# --- MLP: двухвыходная сеть t -> (x, y) ---
N_MLP_C = 200
mlp_c = MLPRegressor(hidden_layer_sizes=(N_MLP_C,), activation="logistic",
                     solver="lbfgs", max_iter=10000, random_state=42, tol=1e-9)
target_xy = np.column_stack([cos_true, sin_true])
mlp_c.fit(t.reshape(-1, 1), target_xy)
mlp_xy = mlp_c.predict(t.reshape(-1, 1))
mlp_x, mlp_y = mlp_xy[:, 0], mlp_xy[:, 1]

# --- Визуализация: 3 subplot ---
fig = make_subplots(
    rows=1, cols=3,
    subplot_titles=(
        "x(t) = cos(t)",
        "y(t) = sin(t)",
        "Окружность в плоскости (x, y)",
    ),
    horizontal_spacing=0.1,
)

# (1) x(t)
fig.add_trace(go.Scatter(x=t, y=cos_true, mode="lines", name="истина cos(t)",
                         line=dict(color="steelblue", width=4)),
              row=1, col=1)
fig.add_trace(go.Scatter(x=t, y=fourier_x, mode="lines", name="Фурье (1 член)",
                         line=dict(color="green", width=2, dash="dot")),
              row=1, col=1)
fig.add_trace(go.Scatter(x=t, y=mlp_x, mode="lines", name=f"MLP ({N_MLP_C} нейронов)",
                         line=dict(color="red", width=2, dash="dash")),
              row=1, col=1)

# (2) y(t)
fig.add_trace(go.Scatter(x=t, y=sin_true, mode="lines",
                         line=dict(color="steelblue", width=4), showlegend=False),
              row=1, col=2)
fig.add_trace(go.Scatter(x=t, y=fourier_y, mode="lines",
                         line=dict(color="green", width=2, dash="dot"), showlegend=False),
              row=1, col=2)
fig.add_trace(go.Scatter(x=t, y=mlp_y, mode="lines",
                         line=dict(color="red", width=2, dash="dash"), showlegend=False),
              row=1, col=2)

# (3) окружность в плоскости (x, y)
fig.add_trace(go.Scatter(x=cos_true, y=sin_true, mode="lines",
                         line=dict(color="steelblue", width=4), showlegend=False),
              row=1, col=3)
fig.add_trace(go.Scatter(x=fourier_x, y=fourier_y, mode="lines",
                         line=dict(color="green", width=2, dash="dot"), showlegend=False),
              row=1, col=3)
fig.add_trace(go.Scatter(x=mlp_x, y=mlp_y, mode="lines",
                         line=dict(color="red", width=2, dash="dash"), showlegend=False),
              row=1, col=3)

fig.update_xaxes(title_text="t", row=1, col=1)
fig.update_xaxes(title_text="t", row=1, col=2)
fig.update_xaxes(title_text="x", row=1, col=3, scaleanchor="y3", scaleratio=1,
                 range=[-1.3, 1.3])
fig.update_yaxes(title_text="x(t)", row=1, col=1)
fig.update_yaxes(title_text="y(t)", row=1, col=2)
fig.update_yaxes(title_text="y", row=1, col=3, range=[-1.3, 1.3])

fig.update_layout(
    title="Окружность: Фурье (1 член) vs MLP (20 нейронов)",
    height=500, template="plotly_white",
)
fig.show()

mse_fourier = np.mean((fourier_x - cos_true) ** 2 + (fourier_y - sin_true) ** 2)
mse_mlp = np.mean((mlp_x - cos_true) ** 2 + (mlp_y - sin_true) ** 2)
print(f"Суммарный MSE (x и y):")
print(f"  Фурье (1 член cos + 1 член sin): {mse_fourier:.2e}")
print(f"  MLP ({N_MLP_C} нейронов):           {mse_mlp:.6f}")
````

### Итог по трём примерам (прямоугольник, треугольник, окружность)

| Задача | Лучший метод | Почему |
|---|---|---|
| Прямоугольник (разрывы) | MLP | Фурье даёт рябь Гиббса у скачков |
| Треугольник (гладкий, с изломами) | Примерно одинаково | Фурье быстро сходится, MLP тоже справляется |
| Окружность (параметрически sin/cos) | **Фурье** | Базис **в точности совпадает** с задачей, MLP учит то же самое через 20 сигмоид |

## Главная мысль

**Нет «лучшего» базиса в вакууме.** Есть подходящие для задачи:
- Периодические гладкие функции (аудио, колебания) → **Фурье**
- Локальные структуры вокруг точки → **Тейлор**
- Произвольные данные, неизвестная форма → **MLP** (учит свой базис сам)

Перцептрон — **универсальный аппроксиматор**, но «универсальность» имеет цену: для конкретной задачи он часто менее эффективен, чем специализированный метод. Преимущество MLP — **не надо знать форму функции заранее**. Именно поэтому нейросети стали доминирующим инструментом в ML: реальные данные (изображения, тексты, временные ряды в бизнесе) не подпадают под простые базисы.

## 11. Мультиклассовая логистическая регрессия — пошаговый разбор на примере чая

Полный проход обучения: softmax, кросс-энтропия, градиенты, обновление весов — на 4 классах китайского чая с 5 признаками.

**Классы:** Тегуань инь (0), Шу пуэр (1), Да Хун Пао (2), Шэн пуэр (3)
**Признаки:** кофеин, теанин, танины, катехины, цвет

Формулы, которые проверим численно:

```
z_k = w_k · x + b_k              # логит класса k
p_k = exp(z_k) / Σ exp(z_j)       # softmax — вероятность класса k
L = −log(p_true)                  # cross-entropy loss
∂L/∂z_k = p_k − y_k               # градиент по логиту
∂L/∂w_k = (p_k − y_k) · x         # градиент по весам
```

````python
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# --- Данные по чаю ---
tea_names = ["Тегуань инь", "Шу пуэр", "Да Хун Пао", "Шэн пуэр"]
feature_names = ["кофеин", "теанин", "танины", "катехины", "цвет"]

X_raw = np.array([
    [3.0, 12, 10, 120, 3],    # Тегуань инь
    [3.5,  6, 20,  40, 9],    # Шу пуэр
    [3.8,  9, 14,  80, 6],    # Да Хун Пао
    [4.2, 11, 19, 140, 4],    # Шэн пуэр
], dtype=float)
y = np.array([0, 1, 2, 3])

print("Исходные данные:")
print(f"{'Чай':<15}" + "".join(f"{f:>10}" for f in feature_names) + f"{'label':>8}")
for i, name in enumerate(tea_names):
    print(f"{name:<15}" + "".join(f"{v:>10.1f}" for v in X_raw[i]) + f"{y[i]:>8d}")

# --- Нормализация ---
mean = X_raw.mean(axis=0)
std = X_raw.std(axis=0)
X = (X_raw - mean) / std

print("\nСтатистика по признакам:")
print(f"{'':<12}" + "".join(f"{m:>10.2f}" for m in mean) + "   <- mean")
print(f"{'':<12}" + "".join(f"{s:>10.2f}" for s in std) + "   <- std")

print("\nНормализованные признаки:")
print(f"{'Чай':<15}" + "".join(f"{f:>10}" for f in feature_names))
for i, name in enumerate(tea_names):
    print(f"{name:<15}" + "".join(f"{v:>+10.2f}" for v in X[i]))
````

### Шаг 1: инициализация W=0, первый прямой проход для Тегуань инь

````python
# Инициализация
K = 4   # число классов
D = 5   # число признаков
W = np.zeros((K, D))
b = np.zeros(K)

def softmax(z):
    ez = np.exp(z - z.max(axis=-1, keepdims=True))  # вычитаем max для численной стабильности
    return ez / ez.sum(axis=-1, keepdims=True)

# Берём первый пример: Тегуань инь
i = 0
x_i = X[i]
y_i = y[i]
y_onehot = np.zeros(K)
y_onehot[y_i] = 1

# Прямой проход
z = W @ x_i + b
p = softmax(z)
L = -np.log(p[y_i])

print(f"Пример: {tea_names[i]}, истинный класс y={y_i}")
print(f"x = {x_i.round(2)}")
print(f"one-hot y = {y_onehot}")
print()
print(f"Логиты z = {z}                  <- все нули, веса пока нулевые")
print(f"Вероятности p = {p.round(3)}       <- равномерное распределение")
print(f"Loss L = -log({p[y_i]:.3f}) = {L:.4f}   <- максимум для K=4: log(4)={np.log(4):.4f}")
````

### Шаг 2: градиенты по формулам ★, ★★, ★★★

````python
# Градиент по логитам
dL_dz = p - y_onehot
print(f"∂L/∂z = p - y = {dL_dz.round(3)}")
print("  ↑ градиент класса 0 отрицательный (надо тянуть z_0 вверх)")
print("  ↑ градиенты остальных положительные (надо тянуть их вниз)")

# Градиент по весам: внешнее произведение (K-вектор) × (D-вектор) = (K × D) матрица
dL_dW = np.outer(dL_dz, x_i)
dL_db = dL_dz

print(f"\n∂L/∂W = (p-y) ⊗ x  — матрица {dL_dW.shape}:")
print(f"{'':<15}" + "".join(f"{f:>10}" for f in feature_names))
for k in range(K):
    print(f"класс {k} ({tea_names[k][:10]:<10})" + "".join(f"{v:>+10.3f}" for v in dL_dW[k]))

print(f"\n∂L/∂b = {dL_db.round(3)}")
````

### Шаг 3: обновление весов (lr=0.5)

````python
lr = 0.5
W_new = W - lr * dL_dW
b_new = b - lr * dL_db

print(f"W_new = W - lr · ∂L/∂W:")
print(f"{'':<15}" + "".join(f"{f:>10}" for f in feature_names))
for k in range(K):
    print(f"класс {k} ({tea_names[k][:10]:<10})" + "".join(f"{v:>+10.3f}" for v in W_new[k]))

print(f"\nb_new = {b_new.round(3)}")

print(f"\nСравним: знаки w_0 и x_0 (Тегуань инь):")
print(f"  w_0_new = {W_new[0].round(3)}")
print(f"  x_0     = {x_i.round(3)}")
print(f"  знаки совпадают → w_0·x_0 будет большим положительным числом")
````

### Шаг 4: проверим, что loss упал

````python
# Пересчитываем предсказание для Тегуань инь
z_new = W_new @ x_i + b_new
p_new = softmax(z_new)
L_new = -np.log(p_new[y_i])

print(f"После обновления:")
print(f"z_new = {z_new.round(3)}        <- класс 0 резко вырос")
print(f"p_new = {p_new.round(3)}  <- уверенность в классе 0 = {p_new[0]*100:.1f}%")
print(f"L_new = {L_new:.4f}              <- было {L:.4f}, упал в {L/L_new:.1f}× раз")
````

### Полное обучение — 200 эпох на всём батче

Теперь обучим модель по-настоящему: на всех 4 примерах одновременно, много эпох.

````python
# Инициализация
W = np.zeros((K, D))
b = np.zeros(K)
lr = 0.3
n_epochs = 200

# One-hot для всех примеров
Y_onehot = np.eye(K)[y]   # (4, 4)

loss_history = []
prob_history = []  # вероятности истинного класса для каждого примера

for epoch in range(n_epochs):
    # Прямой проход по всему батчу
    Z = X @ W.T + b           # (4, 4) — логиты
    P = softmax(Z)            # (4, 4) — вероятности

    # Loss: средний по примерам
    L = -np.mean(np.log(P[np.arange(4), y]))
    loss_history.append(L)
    prob_history.append(P[np.arange(4), y].copy())

    # Градиенты
    dZ = P - Y_onehot         # (4, 4)
    dW = dZ.T @ X / 4         # (K, D) — усреднение по батчу
    db = dZ.mean(axis=0)

    # Обновление
    W -= lr * dW
    b -= lr * db

print(f"Финальный loss: {loss_history[-1]:.6f}")
print(f"\nФинальные вероятности истинного класса:")
for i, name in enumerate(tea_names):
    print(f"  {name:<15}: p(y={y[i]}) = {P[i, y[i]]*100:.2f}%")

print(f"\nФинальная матрица вероятностей (строки — примеры, столбцы — классы):")
print(f"{'':<15}" + "".join(f"{n[:8]:>10}" for n in tea_names))
for i, name in enumerate(tea_names):
    marks = "".join(f"{P[i,k]*100:>9.1f}%" for k in range(K))
    print(f"{name:<15}{marks}  {'<- правильно' if np.argmax(P[i]) == y[i] else '<- ОШИБКА'}")
````

### Визуализация обучения

````python
# График 1: падение loss
fig1 = go.Figure()
fig1.add_trace(go.Scatter(x=list(range(n_epochs)), y=loss_history,
                          mode="lines", line=dict(color="red", width=2),
                          name="Loss"))
fig1.update_layout(title="Loss в процессе обучения (кросс-энтропия)",
                   xaxis_title="эпоха", yaxis_title="L",
                   yaxis_type="log", height=400, template="plotly_white")
fig1.show()

# График 2: рост вероятности истинного класса
prob_hist = np.array(prob_history)  # (n_epochs, 4)
fig2 = go.Figure()
colors = ["#2ecc71", "#8b4513", "#e74c3c", "#f1c40f"]
for i, name in enumerate(tea_names):
    fig2.add_trace(go.Scatter(x=list(range(n_epochs)), y=prob_hist[:, i],
                              mode="lines", name=name,
                              line=dict(color=colors[i], width=2)))
fig2.add_hline(y=0.25, line=dict(color="gray", dash="dash"),
               annotation_text="случайный (0.25)")
fig2.update_layout(title="Вероятность истинного класса для каждого чая во время обучения",
                   xaxis_title="эпоха", yaxis_title="p(y=истина)",
                   height=450, template="plotly_white",
                   yaxis=dict(range=[0, 1.05]))
fig2.show()
````

### Визуализация итоговых весов

Каждый класс «выучил» свой профиль признаков. Покажем матрицу `W` как heatmap.

````python
fig = go.Figure(data=go.Heatmap(
    z=W, x=feature_names, y=tea_names,
    colorscale="RdBu_r", zmid=0,
    text=[[f"{v:+.2f}" for v in row] for row in W],
    texttemplate="%{text}", textfont={"size": 12},
    colorbar=dict(title="вес"),
))
fig.update_layout(
    title="Матрица весов W (строки — классы, столбцы — признаки)<br>"
          "<sup>Красный = положительный вес (признак усиливает класс), синий = отрицательный</sup>",
    height=450, template="plotly_white",
    xaxis=dict(side="top"),
)
fig.show()

print("\nИнтерпретация (читаем по строкам):")
for k in range(K):
    top_pos = feature_names[np.argmax(W[k])]
    top_neg = feature_names[np.argmin(W[k])]
    print(f"  {tea_names[k]:<15}: сильнее всего УВЕЛИЧИВАЕТ логит — {top_pos:>10}, "
          f"сильнее всего УМЕНЬШАЕТ — {top_neg}")
````

### Итог

Прошли полный цикл мультиклассовой логистической регрессии:

1. **Forward:** `z = W·x + b` → `p = softmax(z)` → `L = −log(p_true)`
2. **Backward:** красивая формула `∂L/∂z = p − y` → `∂L/∂w_k = (p_k − y_k)·x`
3. **Обновление:** `w_k ← w_k − lr · ∂L/∂w_k` одновременно для всех классов
4. **Нормализация признаков** обязательна — иначе признаки с большой шкалой доминируют
5. **Регуляризация** `λ·||w||²` предотвращает переобучение (веса не разрастаются)
6. **Сходимость:** loss падает экспоненциально, все 4 класса получают уверенность ~100%
7. **Интерпретируемость:** матрица `W` показывает, какие признаки каждый класс любит / не любит

Этот же принцип работает в нейросетях: последний слой — обычно softmax + cross-entropy, формулы градиентов те же. Вся разница — **сколько слоёв перед softmax** и **какие активации** между ними.

## Итог

1. **Производная** показывает, куда и насколько круто функция растёт/убывает в точке
2. **Градиент** — обобщение производной на много переменных, это вектор
3. **Градиентный спуск** идёт в сторону антиградиента, делая шаг пропорциональный величине градиента
4. **Около минимума** градиент становится маленьким → шаги автоматически замедляются
5. **Learning rate** критичен: слишком маленький — долго, слишком большой — расходится
